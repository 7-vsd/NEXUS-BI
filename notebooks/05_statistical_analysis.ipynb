### Notes on inference
* These are **observational** associations; none of the tests establishes causation.
* With n ≈ 100k, p-values are near zero for trivial effects – read the effect-size column.
* H1 (late → lower reviews) is consistent with a causal story but would require an experiment (e.g. randomised delivery-promise buffers) to quantify the causal impact.
